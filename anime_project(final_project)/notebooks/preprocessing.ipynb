```python
# PREPROCESSING
# Подготовка изображений:
# загрузка
# анализ
# изменение размера
# сохранение

from pathlib import Path
from PIL import Image
```

```python
# Пути к исходному и подготовленному датасетам
ROOT = Path("..")

RAW_PATH = ROOT / "data" / "raw"
PROCESSED_PATH = ROOT / "data" / "processed"

# Создаем папку для подготовленных данных, если она отсутствует
PROCESSED_PATH.mkdir(exist_ok=True)

# Проверяем существование директорий
print(f"RAW: {RAW_PATH.exists()}")
print(f"PROCESSED: {PROCESSED_PATH.exists()}")
```

```python
# Подсчет количества изображений в каждом классе.

def count_images(root_path):
    """
    Подсчитывает количество файлов в каждой папке датасета.

    Args:
        root_path (Path): Путь к корневой директории датасета.

    Returns:
        dict: Словарь вида {"название_класса": количество_изображений}.
    """
    stats = {}

    for folder in root_path.iterdir():
        if folder.is_dir():
            stats[folder.name] = len(list(folder.iterdir()))

    return stats


# Получаем статистику по классам и выводим результат
stats = count_images(RAW_PATH)
print(stats)
```

```python
def preprocess_dataset(source, destination):
    """
    Подготавливает изображения датасета к дальнейшей обработке.

    Для каждого класса создается соответствующая папка в директории назначения,
    после чего изображения конвертируются в формат RGB и сохраняются без
    изменения имени файла.

    Args:
        source (Path): Путь к исходному датасету.
        destination (Path): Путь для сохранения обработанных изображений.

    Returns:
        int: Количество успешно обработанных изображений.
    """
    processed = 0

    for anime_folder in source.iterdir():
        if not anime_folder.is_dir():
            continue

        # Создаем папку для текущего класса
        save_folder = destination / anime_folder.name
        save_folder.mkdir(exist_ok=True)

        for img_path in anime_folder.iterdir():
            try:
                # Загружаем изображение и приводим его к RGB
                img = Image.open(img_path).convert("RGB")

                # Сохраняем изображение в новую директорию
                img.save(save_folder / img_path.name)
                processed += 1

            # Пропускаем поврежденные или неподдерживаемые файлы
            except Exception:
                pass

    return processed


# Запускаем предварительную обработку датасета
total = preprocess_dataset(RAW_PATH, PROCESSED_PATH)

print(f"Обработано: {total}")
```

```python
# Проверяем размер одного обработанного изображения

img_path = next(PROCESSED_PATH.rglob("*.jpg"))

print(Image.open(img_path).size)
```

```python
# Сравниваем количество изображений до и после предварительной обработки

raw_stats = count_images(RAW_PATH)
processed_stats = count_images(PROCESSED_PATH)

print(f"RAW: {sum(raw_stats.values())}")
print(f"PROCESSED: {sum(processed_stats.values())}")
```
