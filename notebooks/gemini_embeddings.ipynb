```python
import os
import sys
from pathlib import Path
sys.path.append(os.path.abspath(".."))
```

```python
from src.embeddings import (
    load_gemini_client,
    generate_gemini_embeddings
)

import json
```

```python
# Пути
ROOT = Path("..")

PROCESSED_PATH = ROOT / "data" / "processed"
EMBEDDINGS_PATH = ROOT / "embeddings"
```

```python
# Настройка эксперимента

# Режим генерации эмбеддингов:
# original — изображения в исходном размере
# 512 — изображения, масштабированные до 512×512
GEMINI_MODE = "original"

SAVE_DIR = EMBEDDINGS_PATH / f"gemini_{GEMINI_MODE}_train"
SAVE_DIR.mkdir(exist_ok=True)
```

```python
# Проверяем доступность необходимых директорий
print(f"Processed: {PROCESSED_PATH.exists()}")
print(f"Save: {SAVE_DIR.exists()}")
```

```python
client = load_gemini_client()
```

```python
print(client.models.list())
```

```python
SPLITS_PATH = ROOT / "data" / "splits"

with open(SPLITS_PATH / "train.json", "r", encoding="utf-8") as f:
    train_images = json.load(f)

print(f"Train images: {len(train_images)}")
```

```python
SAVE_STEP = 100

client = load_gemini_client()

embeddings, labels, paths = generate_gemini_embeddings(
    train_images=train_images,
    save_dir=SAVE_DIR,
    client=client,
    mode=GEMINI_MODE,
    save_step=SAVE_STEP
)
```

```python
print("Embeddings:", len(embeddings))
print("Labels:", len(labels))
print("Paths:", len(paths))
```
